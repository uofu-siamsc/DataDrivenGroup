## 1. Built-in Python

Python can do arithmetic and print text without any extra setup. In a code cell, the value of the **last line** is shown automatically.

```python
3 + 4 * 2
```

To show more than one thing, use `print`.

**Heads up:** powers use `**`, not `^`. `2 ^ 3` is valid Python, but it means something else and gives `1`!

```python
print("Hello, world!")
print(7 / 2)    # division
print(2 ** 10)  # 2 to the 10th power
```

## 2. Variables stick around (so be careful!)

Use `=` to store a value in a **variable**.

```python
x = 5
y = 10
print(x + y)
```

Once a cell has run, its variables stay in the notebook's memory (the **kernel**). Every other cell can use them, even cells *above* it. What matters is the order you **ran** the cells, not the order they appear on the page.

Run the next cell a few times and watch `x` change:

```python
x = x + 1
print(x)
```

Each run starts from whatever `x` was last time. Deleting or editing a cell doesn't erase values that are already in memory either.

**When things get confusing**, click **Restart** at the top of the notebook (this clears the memory), then **Run All** to run every cell from top to bottom.

```python
print(np.sqrt(16))
print(np.pi)
print(np.sin(np.pi / 2))
```

## 4. The "dot"

The dot means "look inside". `np.sqrt` is "the `sqrt` function that lives inside `np`".

Every value in Python is an **object** of some **class** (type), and the class decides what functions come attached to it. These attached functions are called **methods**, and you call them with a dot too. For example, text (a "string") has an `upper` method:

```python
name = "data-driven modeling"
print(type(name))
print(name.upper())
```

A NumPy array (a list of numbers, like a vector) has methods for math. Some things after a dot are stored values (**attributes**) rather than methods. Attributes don't use parentheses.

```python
v = np.array([3, 1, 4, 1, 5])
print(type(v))
print(v.sum())   # method: add up the entries
print(v.max())   # method: the largest entry
print(v.shape)   # attribute: the size of the array (no parentheses)
print(v[0])      # the first entry: Python starts counting at 0, not 1!
```

## 5. Getting help

You don't need to memorize any of this. Here are three ways to find out what's available:
- **Tab completion:** in a code cell, type `np.` or `v.` and press `Tab` (or just wait a moment). VS Code lists everything inside. Hover over a function name to see what it does.
- **`help()`:** prints the documentation for a function, including what inputs it takes and examples. (In Jupyter, `np.linspace?` does the same thing.)
- **Search online:** for example, "numpy linspace". The official documentation is usually the first result.

```python
help(np.linspace)
```

Now let's use `linspace` to make a plot. Notice that this reuses the name `x`, which replaces the number we stored in `x` earlier.

```python
x = np.linspace(0, 2 * np.pi, 100)  # 100 evenly spaced points from 0 to 2*pi

plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")
plt.xlabel("x")
plt.legend()
plt.show()
```
